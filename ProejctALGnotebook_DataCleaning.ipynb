```python
empty = trainset.isnull() 
empcheck = empty.sum()
print(" Columns with missing values")
print(empcheck > 1 ,"\n" )
print(" Number of missing values per column: ")
print(empcheck)
```

```python
#Filling the empty cells with the average values of their individual features 
columns = [
    "CreditScore",
    "Age",
    "Tenure",
    "Balance",
    "NumOfProducts",
    "EstimatedSalary"
]

average_values = trainset[columns].mean().round(2)
trainset[columns] = trainset[columns].fillna(average_values)

print(trainset[columns].isnull().sum())
```

```python
columns2 = [
    "Geography",
    "Gender",
    "HasCrCard",
    "IsActiveMember",
]

replacement_values = {
    "Geography": "France",
    "Gender": "Male",
    "HasCrCard": 1.0,
    "IsActiveMember": 0.0
}

trainset[columns2] = trainset[columns2].fillna(replacement_values)
trainset[columns2].isnull().sum()
trainset.isnull().sum()
```
